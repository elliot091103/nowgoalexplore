```python
# Execute the extraction script to create all views
with open('sql/player_season_tech_extract.sql', 'r', encoding='utf-8') as f:
    sql_script = f.read()
for stmt in sql_script.split(';'):
    if stmt.strip():
        con.sql(stmt)
# -----------------------------------------------------------------------------
# 1. Teams Lookup
# -----------------------------------------------------------------------------
print("=" * 70)
print("1. Tournament Teams (v_player_season_teams)")
print("=" * 70)
con.sql("SELECT * FROM v_player_season_teams").show()
# -----------------------------------------------------------------------------
# 2. Player Profiles
# -----------------------------------------------------------------------------
print("=" * 70)
print("2. Player Profiles (v_player_season_profiles - First 5)")
print("=" * 70)
con.sql("SELECT player_id, player_name, team_id, team_name FROM v_player_season_profiles LIMIT 5").show()
# -----------------------------------------------------------------------------
# 3. Raw 57 Technical Metrics Across Total, Home, Away
# -----------------------------------------------------------------------------
print("=" * 70)
print("3. Technical Stats (v_player_season_stats - First 5)")
print("=" * 70)
con.sql("""
SELECT 
    venue_scope, player_id, matches_played, minutes_played, 
    total_goals, assists, xg, xa, avg_rating, pass_accuracy_pct
FROM v_player_season_stats 
LIMIT 5
""").show()
# -----------------------------------------------------------------------------
# 4. Top Goalscorers Leaderboard (Joined Analytical View)
# -----------------------------------------------------------------------------
print("=" * 70)
print("4. Top Goalscorers Leaderboard (v_player_season_tech_full)")
print("=" * 70)
con.sql("""
SELECT 
    venue_scope,
    player_name,
    team_name,
    matches_played,
    minutes_played,
    total_goals,
    assists,
    xg,
    avg_rating,
    shot_conversion_pct
FROM v_player_season_tech_full 
WHERE venue_scope = 'Total'
ORDER BY total_goals DESC, assists DESC 
LIMIT 10
""").show()
```

```python
# Execute the extraction script to create all views
with open('sql/player_match_tech_extract.sql', 'r', encoding='utf-8') as f:
    sql_script = f.read()
for stmt in sql_script.split(';'):
    if stmt.strip():
        con.sql(stmt)
# -----------------------------------------------------------------------------
# 1. Match Response Envelope & State
# -----------------------------------------------------------------------------
print("=" * 70)
print("1. Match State (v_match_meta)")
print("=" * 70)
con.sql("SELECT * FROM v_match_meta").show()
# -----------------------------------------------------------------------------
# 2. Individual Player Technical Stats (All 35 Metrics)
# -----------------------------------------------------------------------------
print("=" * 70)
print("2. Top Rated Players (v_match_player_tech - Top 5)")
print("=" * 70)
con.sql("""
SELECT 
    team_side, 
    jersey_number, 
    player_name, 
    position_role, 
    rating, 
    shots, 
    shots_on_target, 
    total_passes, 
    pass_accuracy_pct,
    key_passes,
    tackles,
    interceptions
FROM v_match_player_tech 
WHERE is_valid 
ORDER BY rating DESC 
LIMIT 5
""").show()
# -----------------------------------------------------------------------------
# 3. Match In-Game Events Log
# -----------------------------------------------------------------------------
print("=" * 70)
print("3. In-Game Events Log (v_match_player_events - First 6)")
print("=" * 70)
con.sql("""
SELECT 
    team_side, 
    jersey_number, 
    player_name, 
    position_role, 
    event_id, 
    event_name 
FROM v_match_player_events 
LIMIT 6
""").show()
# -----------------------------------------------------------------------------
# 4. Team Technical Summary (Home vs Away Comparison)
# -----------------------------------------------------------------------------
print("=" * 70)
print("4. Team Technical Comparison (v_match_team_stats_summary)")
print("=" * 70)
con.sql("""
SELECT 
    team_side,
    players_featured,
    team_avg_rating,
    total_shots,
    total_shots_on_target,
    shot_accuracy_pct,
    total_passes,
    team_pass_accuracy_pct,
    total_key_passes,
    total_tackles,
    total_interceptions,
    total_fouls
FROM v_match_team_stats_summary
""").show()
```

```python
odds_scripts = [
    "sql/match_odds_3in1_extract.sql",
    "sql/match_odds_corners_extract.sql",
    "sql/match_odds_doubleschance_extract.sql",
    "sql/match_odds_eurohandicap_extract.sql"
]
# Execute all odds extraction scripts to create the views
for script in odds_scripts:
    with open(script, 'r', encoding='utf-8') as f:
        sql = f.read()
    for stmt in sql.split(';'):
        if stmt.strip():
            con.sql(stmt)
print("=" * 80)
print("1. 3in1 Asian Handicap (v_odds_3in1_asian_handicap - Recent 5)")
print("=" * 80)
con.sql("""
SELECT 
    modify_time, match_minute, home_score, guest_score, 
    home_odds, handicap_line, away_odds, is_closed, stage_type
FROM v_odds_3in1_asian_handicap 
WHERE home_odds > 0 
LIMIT 5
""").show()
print("=" * 80)
print("2. 3in1 1X2 European Odds (v_odds_3in1_1x2 - Recent 5)")
print("=" * 80)
con.sql("""
SELECT 
    modify_time, match_minute, home_score, guest_score, 
    home_win_odds, draw_odds, away_win_odds, is_closed, stage_type
FROM v_odds_3in1_1x2 
WHERE home_win_odds > 0 
LIMIT 5
""").show()
print("=" * 80)
print("3. 3in1 Over/Under Goals (v_odds_3in1_over_under - Recent 5)")
print("=" * 80)
con.sql("""
SELECT 
    modify_time, match_minute, home_score, guest_score, 
    over_odds, goal_line, under_odds, is_closed, stage_type
FROM v_odds_3in1_over_under 
WHERE over_odds > 0 
LIMIT 5
""").show()
print("=" * 80)
print("4. Corners Over/Under (v_odds_corners_over_under - Recent 5)")
print("=" * 80)
con.sql("""
SELECT 
    modify_time, match_minute, home_corners, guest_corners, total_corners, 
    corner_over_odds, corner_line, corner_under_odds, stage_type
FROM v_odds_corners_over_under 
WHERE corner_over_odds IS NOT NULL 
LIMIT 5
""").show()
print("=" * 80)
print("5. Double Chance & Euro Handicap Views")
print("=" * 80)
con.sql("SELECT * FROM v_odds_double_chance").show()
con.sql("SELECT * FROM v_odds_euro_handicap").show()
```
